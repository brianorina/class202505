# MACHINE LEARNING - BREAST CANCER DIAGNOSIS

# 1 - DEFINE THE PROBLEM

Breast cancer is one of the most common types of cancer among women worldwide.  
Early diagnosis is crucial to improving the chances of successful treatment and survival. 

The objective of this project is to develop a machine learning model capable of predicting whether a tumor is malignant or benign based on various features computed from a digitized image of a fine needle aspirate (FNA) of a breast mass.

The desired output is a binary classification:  
- `Malignant (M)` or  
- `Benign (B)`

This is a **supervised learning** problem and the task is **binary classification**.

# 7 - SPLIT THE DATA

```python
# Split the dataset into features (X) and target (y).
# Exclude both 'target' (the label) and 'diagnosis' (the string label) from the features.
X = df.drop(['target', 'diagnosis'], axis=1)
y = df['target']

# Split the data into training and test sets.
# 30% of the data is reserved for testing and a fixed random state ensures reproducibility.
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

# 8 - CREATE & TRAIN A MODEL

```python
# Create an instance of the logistic regression model.
# The maximum number of iterations is set to 10,000 to ensure convergence.
model = LogisticRegression(max_iter=10000)
```

```python
# Train (fit) the logistic regression model using the training data.
model.fit(X_train, y_train)
```

# 9 - EVALUATE AND TUNE THE MODEL

```python
# Make predictions on the test set using the trained model.
y_pred = model.predict(X_test)

# Evaluate model
accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)
print(f'Accuracy: {accuracy:.2f}')
print('Confusion Matrix:')
print(conf_matrix)
```

```python
# Calculate the precision score for the model's predictions.
precision = precision_score(y_test, y_pred)
print(f'Precision: {precision:.2f}')
```
